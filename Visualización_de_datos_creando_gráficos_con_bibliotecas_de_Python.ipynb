# Conociendo la biblioteca Matplotlib

```python
import pandas as pd
```

```python
df = pd.read_csv("/content/inmigrantes_canada.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df = df.set_index("Pais")
df
#convertimos una columna en indice para trabajar mejor donde cada registro se asocia a cada valores de Pais
#cuando usemos loc ya no tendremos que poner filtro para buscar una fila especifica
```

```python
años = list(map(str,range(1980,2014)))
años
```

```python
colombia = df.loc["Colombia",años]
colombia
#filtramos por fila : buscamos el valor de "Colombia" en filas, como froma parte de un indice, solo buscamos el valor
#    de lo contrario tendriamos : df["Pais"] == "Colombia"
#filtramos por columna : lista de años (nombres de las columnas)
#nos mostrara una serie
#    pandas al seleccionar una sola fila --> hace una transpuesta : las columnas ahora son indices y el indice "Colombia" sera columna en
#    esa serie
```

```python
colombia.values
```

```python
colombia.values.tolist()
```

```python
col_dict = {
    "Año":colombia.index.tolist(),
    "Inmigrantes":colombia.values.tolist()
    }
col_dict
#colombia.index : objeto donde muestra todos los indices de la serie (años)
#tolist() : convierte a una lista, optimo, metodo de pandas

#colombia.values : objeto donde muestra todos los valores de la columna de la serie (Colombia, inmigrantes), valores estan en np.int64
#tolist() : convierte a una lista, optimo, metodo de pandas
```

Convertimos en un DataFrame nuestro diccionario que tiene los datos de nuestra serie colombia

```python
datos_col = pd.DataFrame(col_dict)
datos_col.head()
#convertimos el dict a un DataFrame
```

Creamos un grafico usando los datos de nuestro DataFrame datos_col


```python
import matplotlib.pyplot as plt
```

```python
ticks_eje_x = list(map(str, range(1980,2014, 5)))  #lista donde estara strings de 19980 hasta 2013 que van de 5 en 5
ticks_eje_x
```

```python
plt.figure(figsize=(8,4))   #tamaño
plt.plot(datos_col["Año"], datos_col["Inmigrantes"])   #datos relacinables valores eje x, eje y
plt.xticks(ticks_eje_x)     #especificar cuales y cuantos años se van a mostar en eje x
plt.show()                  #muestra el grafico
```

```python
plt.figure(figsize=(8,4))   #tamaño
plt.plot(datos_col["Año"], datos_col["Inmigrantes"])   #datos relacinables valores eje x, eje y
plt.xticks(ticks_eje_x)     #especificar cuales y cuantos años se van a mostar en eje x
plt.title("Inmigracionn de Colombia hacia Canada")     #titulo
plt.xlabel("Año")                                      #etiqueta para eje x
plt.ylabel("Numero de inmigrantes")                    #etiqueta para eje y
plt.show()                  #muestra el grafico, tiene que estar al ultimo para que ejecute todo lo de arriba
```

# Creando una figura

vamos a guardar un lienzo con su grafica en una variable (fig : lienzo)
- lienzo contiene el grafica, asi que la variable que contiene todo es "fig"

```python
fig, ax = plt.subplots(figsize=(8,4))   #fig : lienzo, ax : imagen, grafica, figura de ejes
ax.plot(datos_col["Año"], datos_col["Inmigrantes"])   #ax : figura ejes le damos los valores
ax.xaxis.set_major_locator(plt.MultipleLocator(5))    #xticks solo va colocar las posiciones 5 en 5
ax.set_title("inmigrantes de colombia hacia Canadá\n1980 - 2013")  #titulo
ax.set_xlabel("Año")                                  #etiqueta eje x
ax.set_ylabel("Numero de inmigrantes")                #etiqueta eje y
plt.show()     #mostramos la grafica ya con los datos
```

```python
fig, axs = plt.subplots(1,2,figsize=(10,4))                  #dos graficas (1: 1 fila, 2: 2 columnas)
#estos son solo variables

axs[0].plot(datos_col["Año"], datos_col["Inmigrantes"])      #añades datos al axs (figura de ejes)
axs[0].set_title("Inmigrantes de Colombia hacia Canadá\n1980 - 2013")    #titulo al axs
axs[0].set_xlabel("Año")                                                 #etiqueta eje x
axs[0].set_ylabel("Numero de inmigrantes")                               #etiqueta eje y
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))       #xticks solo va colocar las posiciones 5 en 5
axs[0].grid()                                                #crea una malla

axs[1].boxplot(datos_col["Inmigrantes"])                     #grafia tipo bocplot (cajita)
axs[1].set_title("Numero de inmigrantes de colombianos\nhacia Canadá 1980 - 2013")   #titulo al axs
axs[1].set_xlabel("Colombia")                                #etiqueta eje x
axs[1].set_ylabel("Numero de inmigrantes")                   #etiqueta eje y
axs[1].grid()                                                #crea una malla

plt.show()                                                   #muestra la grafica
```

```python
datos_col.describe()   #muestra datos estadisticos que el bloxplot representa (algunos)
#boxplot explicacion
# linea inferior : min
# linea superior : max
# cajita
#    linea inferior : 25% (primer quartil) : el 25% datos son menores igual a 372 (datos ordenados)
#    linea naranja : 50% (segundo quartil) : el 50 % datos son menores igual a 633
#    linea superior : 75% (tercer quartil) : el 75 % datos son menores igual a 4173
#    los otros 25% son los datos restantes : mayores al 75%, mayor que 4173
```

```python
df.loc["Argentina", años]   #serie, indice : años (implicito), columna : Argentina
#cuando ponemos esta serie en un plt.plot(), el indice es el eje x, y los valores columna es eje y
```

```python
fig, axs = plt.subplots(2,2,figsize=(10,6))   #lienzo con 2 filas, 2 columnas (4 graficas)
fig.subplots_adjust(hspace=0.5, wspace=0.3)  #espacio entre graficas, en pulgadas, por defecto es 0.2
#hspace : espacio vertical entre graficas
#wspace : espacio horizontal entre grafica
#valores : 0.5 significa que el espaciado será del 50% de la altura/anchura de la figura.


fig.suptitle("Inmigracion de los 4 mayores paises sudamericanos\nhacia Canada desde 1980 hasta 2013")

axs[0,0].plot(df.loc["Colombia", años])
axs[0,0].set_title("Colombia")

axs[0,1].plot(df.loc["Brasil", años])
axs[0,1].set_title("Brasil")

axs[1,0].plot(df.loc["Argentina", años])
axs[1,0].set_title("Argentina")

axs[1,1].plot(df.loc["Perú", años])
axs[1,1].set_title("Perú")

for ax in axs.flat:                                       #axs.flat : acceder a cada una de la graficas
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))    #grafica i los x ticks mostrar valores 5 en 5
    ax.set_xlabel("Año")                                  #grafica i etiqueta eje x
    ax.set_ylabel("Numero de inmigrantes")                #grafica i etiqueta eje y
    ax.grid()

ymin = 0
ymax = 7000

for ax in axs.ravel():       #ravel() : array multidimensional (lienzo 2x2) a unidimensional (lista 4 graficas)
  ax.set_ylim(ymin, ymax)    #set_ylim : poner limites al valores eje y  (min, max), itera a las 4 graficas
                             #para asi tener todos los graficos misma escala

plt.show()
```
